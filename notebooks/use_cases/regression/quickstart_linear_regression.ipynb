## Load and inspect the data

Each row is a cardholder observed in one quarter. The target, `Balance`, is the average monthly revolving balance in dollars; customers who pay in full each month have a balance of zero.

| Column | Description |
|---|---|
| `Quarter` | Quarter in which the customer was observed |

| `DebtToIncome` | Monthly debt payments as a share of income |
| `CreditScore` | Bureau-style credit score, 300–850 |
| `CreditLimit` | Credit limit on the card, in dollars |
| `RecentInquiries` | Credit inquiries in the last six months |

| `Region` | Region of residence |

## Develop the champion model

The champion is an ordinary least squares (OLS) regression fitted with `statsmodels`. We use the formula API, which keeps the variable names and handles categorical variables such as `Region`.

<a id='toc5_1__'></a>

### Fit a candidate model

Start with every candidate variable, fitted on the development data as received:

````python
CANDIDATE_FORMULA = (
    "Balance ~ Income + DebtToIncome + CreditScore + CreditLimit + RecentInquiries"
    " + NumCards + TenureYears + IsStudent + C(Region)"
)

candidate_fit = smf.ols(CANDIDATE_FORMULA, data=development_df).fit()
candidate_fit.params.round(3)
````

Before running any test, read the coefficients against business sense. A higher credit limit should mean a higher balance, and with the limit held fixed, a higher income a lower one. Here `CreditLimit` and `Income` do not behave that way — a sign that something in the data is distorting the fit. Let's find out what.

<a id='toc5_2__'></a>

### Investigate influential observations

Initialize the candidate as a ValidMind model and run the `InfluentialObservations` test, which flags rows that individually move the estimates:

````python
vm_candidate_model = vm.init_model(candidate_fit, input_id="candidate_model")

vm.tests.run_test(
    "validmind.model_validation.statsmodels.InfluentialObservations",
    inputs={"model": vm_candidate_model},
)
````

Look at the observations above the severe Cook's distance threshold, and at the coefficient table beside them: each of those rows, on its own, moves the estimates materially. The test assesses observations one at a time, and one extreme row can mask others like it — so inspect the flagged rows, then search the data for the same pattern. The source system records incomes up to a fixed cap, so any income above it is an error:

````python
INCOME_CAP = 200  # $000s: the source system records incomes up to $200k

suspicious = raw_df[raw_df["Income"] > INCOME_CAP]
suspicious
````

These incomes are about ten times a plausible value, while the same customers' balances, limits and scores are typical: they are data entry errors (a misplaced decimal point), not genuine high earners. We remove them **before** splitting the data, so that no error reaches the test or out-of-time samples either.

````python
clean_df = raw_df[raw_df["Income"] <= INCOME_CAP]
train_df, test_df, oot_df = ccb.preprocess(clean_df)

print(f"Train {len(train_df):,} | Test {len(test_df):,} | Out of time {len(oot_df):,}")
````

<a id='toc5_3__'></a>

### Select the variables

Refit the candidate on the clean training data and check multicollinearity with the `VarianceInflationFactor` test:

````python
clean_candidate = vm.init_model(
    smf.ols(CANDIDATE_FORMULA, data=train_df).fit(), input_id="clean_candidate"
)

vm.tests.run_test(
    "validmind.model_validation.statsmodels.VarianceInflationFactor",
    inputs={"model": clean_candidate},
)
````

`CreditScore` and `CreditLimit` are flagged as severely collinear: they carry almost the same information, because card limits are set largely from the credit score. We keep `CreditLimit`, which drives balances directly, and drop `CreditScore`.

Next, check which variables add nothing — coefficients not statistically different from zero. Look at `TenureYears` and the `Region` dummies:

````python
clean_candidate.model.pvalues.drop("Intercept").round(3).sort_values(ascending=False)
````

The champion keeps the variables with a clear, significant effect:

````python
FEATURES = ["Income", "DebtToIncome", "CreditLimit", "RecentInquiries", "NumCards", "IsStudent"]
CHAMPION_FORMULA = "Balance ~ " + " + ".join(FEATURES)

champion_fit = smf.ols(CHAMPION_FORMULA, data=train_df).fit()
champion_fit.summary()
````

<a id='toc6__'></a>

## Build a challenger model

A linear model of a balance has a structural weakness: balances cannot be negative, and many customers have none at all, but a straight line will predict negative values for low-risk customers. The challenger addresses this with a **two-part model**, a standard approach for targets with many zeros:

1. a logistic regression for whether the customer carries a balance, and
2. an OLS regression for how much, fitted on customers who do.

Its prediction is the probability of carrying a balance times the expected amount. ValidMind passes either a DataFrame or an array to `predict`, so the class accepts both:

````python
class TwoPartModel:
    """P(balance > 0) from a logit, times the expected balance when positive from OLS."""

    def __init__(self, formula_rhs, features, train):
        self.features = features
        has_balance = train.assign(HasBalance=(train["Balance"] > 0).astype(int))
        self.probability = smf.logit(f"HasBalance ~ {formula_rhs}", data=has_balance).fit(disp=0)
        self.amount = smf.ols(f"Balance ~ {formula_rhs}", data=train[train["Balance"] > 0]).fit()

    def predict(self, X):
        X = pd.DataFrame(X, columns=self.features)
        expected_amount = self.amount.predict(X).clip(lower=0)
        return np.asarray(self.probability.predict(X) * expected_amount)


challenger = TwoPartModel(" + ".join(FEATURES), FEATURES, train_df)
````

<a id='toc7__'></a>

## Initialize the ValidMind inputs

Before running tests, register the models and datasets with ValidMind. Each gets an `input_id`, which is how test results — and the template's configuration — refer to it.

<a id='toc7_1__'></a>

### Initialize the models

`vm.init_model` wraps a trained model so tests can use it. It recognizes the library the model comes from — here, a fitted `statsmodels` result for the champion, and the custom `TwoPartModel` class for the challenger, which ValidMind treats like a scikit-learn model with a `predict` method:

````python
vm_champion = vm.init_model(champion_fit, input_id="champion")
vm_challenger = vm.init_model(challenger, input_id="challenger")
````

<a id='toc7_2__'></a>

### Initialize the datasets

`vm.init_dataset` wraps a DataFrame and tells ValidMind how to read it:

- `target_column` — the column the model predicts,
- `feature_columns` — the columns the model uses,
- `extra_columns` — columns the model does *not* use but tests can still look at. Here, `CreditScore`, `TenureYears`, `Region` and `Quarter` are attached this way so the validation can segment results by them. Looking for bias along variables the model does not use is one of the most revealing checks a validator can run.

We create one dataset per sample, plus the raw data as received for the data-quality tests, and the test and out-of-time samples together for tracking performance quarter by quarter. Each gets a copy of its DataFrame, because ValidMind adds prediction columns to the data it is given:

````python
EXTRA_COLUMNS = {column: None for column in ["CreditScore", "TenureYears", "Region", "Quarter"]}

vm_raw = vm.init_dataset(
    raw_df.copy(),
    input_id="raw_data",
    target_column=ccb.target_column,
    feature_columns=ccb.feature_columns,
    extra_columns={ccb.time_column: None},
)

vm_train = vm.init_dataset(
    train_df.copy(),
    input_id="train_dataset",
    target_column=ccb.target_column,
    feature_columns=FEATURES,
    extra_columns=EXTRA_COLUMNS,
)

vm_test = vm.init_dataset(
    test_df.copy(),
    input_id="test_dataset",
    target_column=ccb.target_column,
    feature_columns=FEATURES,
    extra_columns=EXTRA_COLUMNS,
)

vm_oot = vm.init_dataset(
    oot_df.copy(),
    input_id="oot_dataset",
    target_column=ccb.target_column,
    feature_columns=FEATURES,
    extra_columns=EXTRA_COLUMNS,
)

test_and_oot_df = pd.concat([test_df, oot_df])
vm_test_and_oot = vm.init_dataset(
    test_and_oot_df.copy(),
    input_id="test_and_oot_dataset",
    target_column=ccb.target_column,
    feature_columns=FEATURES,
    extra_columns=EXTRA_COLUMNS,
)
````

<a id='toc7_3__'></a>

### Assign predictions

`assign_predictions` links a model's predictions to a dataset. Tests then compare actual and predicted values directly, without calling the model again, and each dataset can hold the predictions of several models side by side.

There are two ways to assign them:

- **Let ValidMind call the model.** Pass only the model, and ValidMind calls its `predict` method on the dataset's features. The challenger works this way.
- **Pass the predictions yourself** with `prediction_values`. Use this when the model needs its input in a particular form, or when predictions were computed elsewhere. ValidMind passes features as an array, and a `statsmodels` formula model needs named columns, so for the champion we compute the predictions with `champion_fit.predict` and pass them in.

Each dataset gets the predictions of the models that are tested on it:

````python
# Champion: predictions computed by statsmodels and passed in
vm_train.assign_predictions(vm_champion, prediction_values=champion_fit.predict(train_df).values)
vm_test.assign_predictions(vm_champion, prediction_values=champion_fit.predict(test_df).values)
vm_oot.assign_predictions(vm_champion, prediction_values=champion_fit.predict(oot_df).values)
vm_test_and_oot.assign_predictions(
    vm_champion, prediction_values=champion_fit.predict(test_and_oot_df).values
)

# Challenger: ValidMind calls its predict method — compared with the champion on the test sample
vm_test.assign_predictions(vm_challenger)
````

## What the validation covers

This quickstart runs the whole validation in one go. The template groups its tests into eight areas, each answering one question a validator asks of a linear regression model:

| Area | The question | Tests |
|---|---|---|
| **Data** | Is the data fit to build and validate the model on? | Data quality, outliers, skewness and correlations, on the data as received |
| **Model specification** | Are the right variables in the model, with sensible and stable effects? | Influential observations, multicollinearity (VIF), coefficients and their expected signs, robust standard errors, bootstrap stability, linearity |
| **Model assumptions** | Do the errors behave as ordinary least squares assumes? | Normality and constant variance of the residuals |
| **Performance and uncertainty** | How accurate is the model on unseen data, and is it honest about its uncertainty? | Accuracy, prediction-interval coverage, impossible predictions, calibration, the tails |
| **Champion and challenger** | Is there a better model, and is the difference real? | Side-by-side errors, and confidence intervals for the difference |
| **Sensitivity and stress testing** | Does the output respond plausibly when the inputs move? | One-at-a-time shocks and joint stress scenarios |
| **Segment analysis** | Where in the portfolio is the model wrong? | Bias and weak spots by segment, including variables the model does not use |
| **Stability over time** | Does the model still work on newer customers? | Out-of-time performance, performance quarter by quarter, population drift |

The tests are of four kinds, and they are read differently:

- **Formal statistical tests** give a p-value and a pass/fail. In large samples they also detect tiny, immaterial effects.
- **Threshold checks** compare a measure with a conventional limit — read the margin, not just the verdict.
- **Diagnostics** are tables and plots that show *where* and *how much* — this is where materiality is judged.
- **Scenarios** project outcomes under assumed conditions, to check the response is plausible.

<a id='toc8_1__'></a>

### Configure and run the tests

The template lists every test with the input it applies to, using a suffix after the test ID — for example `PredictionIntervals:test` and `PredictionIntervals:oot`. The configuration below gives each test its inputs, and the parameters only a validator can supply: the **expected sign** of each coefficient, the **feasible range** of a balance (it cannot be negative), and the **stress scenarios** to project.

Durbin-Watson is deliberately not used: it tests autocorrelation in row order, which has no meaning for customers shuffled into a training sample. Whether errors drift over time is tested directly instead, in the stability area.

````python
T = "validmind.model_validation"

EXPECTED_SIGNS = {
    "Income": "-",  # with the limit held fixed, higher earners pay down more
    "DebtToIncome": "+",
    "CreditLimit": "+",
    "RecentInquiries": "+",
    "NumCards": "+",
    "IsStudent": "+",
}
SCENARIOS = {
    "Recession": {"Income": {"multiply": 0.8}, "DebtToIncome": {"add": 0.05}},
    "Credit line increase": {"CreditLimit": {"multiply": 1.1}},
    "Severe stress": {
        "Income": {"multiply": 0.7},
        "DebtToIncome": {"add": 0.10},
        "RecentInquiries": {"add": 2},
    },
}

raw = {"inputs": {"dataset": vm_raw}}
champion_on_test = {"inputs": {"dataset": vm_test, "model": vm_champion}}
challenger_on_test = {"inputs": {"dataset": vm_test, "model": vm_challenger}}

config = {
    # Data
    "validmind.data_validation.DatasetDescription:raw_data": raw,
    "validmind.data_validation.DescriptiveStatistics:raw_data": raw,
    "validmind.data_validation.MissingValues:raw_data": raw,
    "validmind.data_validation.Duplicates:raw_data": raw,
    "validmind.data_validation.IQROutliersTable:raw_data": raw,
    "validmind.data_validation.Skewness:raw_data": raw,
    "validmind.data_validation.PearsonCorrelationMatrix:raw_data": raw,
    "validmind.data_validation.HighPearsonCorrelation:raw_data": raw,
    "validmind.data_validation.DatasetSplit": {
        "inputs": {"datasets": [vm_train, vm_test, vm_oot]}
    },
    # Model specification
    f"{T}.statsmodels.InfluentialObservations:candidate_model": {
        "inputs": {"model": vm_candidate_model}
    },
    f"{T}.statsmodels.VarianceInflationFactor:candidate_model": {
        "inputs": {"model": clean_candidate}
    },
    f"{T}.statsmodels.VarianceInflationFactor:champion": {"inputs": {"model": vm_champion}},
    f"{T}.statsmodels.RegressionCoeffs:champion": {"inputs": {"model": vm_champion}},
    f"{T}.statsmodels.ExpectedCoefficientSigns:champion": {
        "inputs": {"model": vm_champion},
        "params": {"expected_signs": EXPECTED_SIGNS},
    },
    f"{T}.statsmodels.RobustStandardErrors:champion": {"inputs": {"model": vm_champion}},
    f"{T}.statsmodels.CoefficientStability:champion": {"inputs": {"model": vm_champion}},
    f"{T}.statsmodels.LinearitySpecification:champion": {"inputs": {"model": vm_champion}},
    # Assumptions, on the data the model was fitted on
    f"{T}.ResidualNormality:champion": {"inputs": {"dataset": vm_train, "model": vm_champion}},
    f"{T}.Heteroscedasticity:champion": {"inputs": {"dataset": vm_train, "model": vm_champion}},
    # Performance and uncertainty, on held-out data
    f"{T}.sklearn.RegressionR2Square:champion_test": champion_on_test,
    f"{T}.sklearn.RegressionPerformance:champion_test": champion_on_test,
    f"{T}.statsmodels.PredictionIntervals:test": champion_on_test,
    f"{T}.PredictionRange:champion_test": {**champion_on_test, "params": {"lower_bound": 0}},
    f"{T}.CalibrationByDecile:champion_test": champion_on_test,
    f"{T}.TailPerformance:champion_test": champion_on_test,
    # Champion and challenger
    f"{T}.sklearn.RegressionErrorsComparison:champion_vs_challenger": {
        "inputs": {"datasets": [vm_test], "models": [vm_champion, vm_challenger]}
    },
    f"{T}.PerformanceConfidenceIntervals:champion_vs_challenger": {
        "inputs": {"dataset": vm_test, "models": [vm_champion, vm_challenger]}
    },
    f"{T}.PredictionRange:challenger_test": {**challenger_on_test, "params": {"lower_bound": 0}},
    # Sensitivity and stress
    f"{T}.FeatureSensitivity:champion": champion_on_test,
    f"{T}.FeatureSensitivity:challenger": challenger_on_test,
    f"{T}.ScenarioAnalysis:champion": {
        **champion_on_test,
        "params": {"scenarios": SCENARIOS, "lower_bound": 0},
    },
    f"{T}.ScenarioAnalysis:challenger": {
        **challenger_on_test,
        "params": {"scenarios": SCENARIOS, "lower_bound": 0},
    },
    # Segments — every column by default, including the extra columns
    f"{T}.ResidualBiasBySegment:champion_test": champion_on_test,
    f"{T}.RegressionWeakspots:champion_test": champion_on_test,
    f"{T}.ResidualBiasBySegment:challenger_test": challenger_on_test,
    # Stability over time
    f"{T}.sklearn.RegressionErrorsComparison:in_time_vs_out_of_time": {
        "inputs": {"datasets": [vm_test, vm_oot], "models": [vm_champion]}
    },
    f"{T}.PerformanceOverTime:champion": {
        "inputs": {"dataset": vm_test_and_oot, "model": vm_champion},
        "params": {"time_column": "Quarter"},
    },
    f"{T}.statsmodels.PredictionIntervals:oot": {
        "inputs": {"dataset": vm_oot, "model": vm_champion}
    },
    # PSI between 0.1 and 0.25 is conventionally a moderate shift worth investigating
    "validmind.ongoing_monitoring.FeatureDrift:development_vs_out_of_time": {
        "inputs": {"datasets": [vm_train, vm_oot]},
        "params": {"psi_threshold": 0.1},
    },
}

full_suite = vm.run_documentation_tests(config=config)
````

## Read the results

The results are now in your model's documentation on the ValidMind Platform, section by section. The synthetic data was built so that a validation has issues to find — here is what to look for in each area, and why it matters:

- **Data.** The tests run on the data *as received*: look for the implausible incomes among the outliers, the strong correlation between `CreditScore` and `CreditLimit`, and the many zero balances that make the target bounded and skewed.
- **Model specification.** Compare the candidate with the champion — the influence and multicollinearity results on the candidate are the evidence for the decisions that produced the champion. In the linearity test, the partial-residual plot for `RecentInquiries` flattens out: a diminishing effect that a straight line overstates.
- **Model assumptions.** With thousands of rows, the formal tests reject over small departures; judge materiality from the residual shape and spread. The zero balances are the usual cause.
- **Performance and uncertainty.** Headline accuracy is solid, but look for predictions of negative balances, and for a calibration curve that bends away from the diagonal at the ends.
- **Champion and challenger.** A difference matters only if its confidence interval excludes zero — and even then it must be weighed against the challenger's added complexity.
- **Sensitivity and stress testing.** Check the direction and size of every effect. The severe-stress projection extrapolates `RecentInquiries` along a straight line.
- **Segment analysis.** `Region` and `TenureYears` have no real effect, so they should show no bias — evidence the test is not raising false alarms. Then look at `CreditScore`, a variable the champion no longer uses: dropping it was sound by every average-based test, and segmenting on it shows what that decision cost.
- **Stability over time.** Look for when bias begins, whether prediction intervals miss on one side, and which inputs drifted — together they separate a change in *who* the customers are from a change in *how* they behave.

### What the validator would conclude

Check each finding against your own run's output before drawing a conclusion.

| Finding | Evidence | Severity | Recommendation |
|---|---|---|---|
| Under-prediction out of time | Stability over time | High | Recalibrate on recent quarters before use, and monitor bias quarterly |
| Lowest credit scores under-predicted | Segment analysis | High | Bring credit risk back into the model in a form that avoids the collinearity with the limit, such as a subprime indicator |
| Negative predictions | Performance and uncertainty; champion and challenger | Medium | Adopt the two-part challenger, or floor predictions at zero with the limitation documented |
| Diminishing effect of inquiries | Model specification; sensitivity and stress testing | Low | Use a transformed term, such as `np.log1p(RecentInquiries)` |
| Non-normal, non-constant errors | Model assumptions | Low | Rely on robust standard errors and backtested intervals rather than analytical ones |

None of these findings is visible in the champion's headline accuracy. They come from interrogating the model by segment, over time, and at the edges of its range.
